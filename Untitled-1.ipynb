# Редукция диффузионной модели к задаче Ширяева для $\mathcal{N}(0,1)$

**Идея:** вместо высокоразмерных наблюдений $X_t \in \mathbb{R}^d$ с неизвестным $p_1$ перейти к скалярной статистике с **явно известными** обоими распределениями.

---

## Шаг 1 — Нормированная ошибка денойзинга

Для каждого $X_t$ вычисляем:

$$L_t = \left\|\hat\varepsilon_\theta\!\left(\mu(\tau)\, X_t + \sigma(\tau)\,\varepsilon,\;\tau,\;\mathbf{h}_{t-1}\right) - \varepsilon\right\|^2, \qquad \varepsilon \sim \mathcal{N}(0, I)$$

По hold-out выборке $X^{(0)} \sim p_0$ оцениваем $\mu_0 = \mathbb{E}[L_t^{(0)}]$, $\sigma_0 = \mathrm{Std}[L_t^{(0)}]$ и нормируем:

$$\boxed{Z_t = \frac{L_t - \mu_0}{\sigma_0}}$$

---

## Шаг 2 — Распределения $Z_t$ известны точно

| Режим | Распределение |
|---|---|
| До разладки $\;(X_t \sim p_0)$ | $Z_t \approx \mathcal{N}(0,\,1)$ |
| После разладки $\;(X_t \sim p_1)$ | $Z_t \approx \mathcal{N}(\delta,\,1)$, где $\delta > 0$ |

**Обоснование $\delta > 0$.** По принципу минимума MSE оптимальный денойзер для $p_1$ отличается от обученного для $p_0$:

$$\mathbb{E}_{p_1}\!\left[\|\hat\varepsilon_\theta - \varepsilon\|^2\right] = \underbrace{\mathbb{E}_{p_1}\!\left[\|\hat\varepsilon_1 - \varepsilon\|^2\right]}_{\text{минимальная ошибка}} + \underbrace{\mathbb{E}_{p_1}\!\left[\|\hat\varepsilon_\theta - \hat\varepsilon_1\|^2\right]}_{= \delta\,\sigma_0 > 0 \text{ при } p_1 \neq p_0}$$

---

## Шаг 3 — Точное отношение правдоподобий

$$\boxed{\ell_t = \log\frac{p_1(Z_t)}{p_0(Z_t)} = \delta\, Z_t - \frac{\delta^2}{2}}$$

$$\Lambda_t = \exp\!\left(\delta\, Z_t - \frac{\delta^2}{2}\right)$$

---

## Шаг 4 — Статистика Ширяева–Робертса

$$\boxed{R_t = (1 + R_{t-1})\cdot\Lambda_t, \qquad R_0 = 0}$$

Апостериорная вероятность разладки (с дискретной интенсивностью $\lambda_d = 1 - e^{-\lambda\Delta t}$):

$$\pi_t = \frac{\left[\pi_{t-1} + (1-\pi_{t-1})\,\lambda_d\right]\Lambda_t}{\left[\pi_{t-1} + (1-\pi_{t-1})\,\lambda_d\right]\Lambda_t + (1-\pi_{t-1})(1-\lambda_d)}$$

$$\pi_t \approx \frac{\lambda_d\, R_t}{1 + \lambda_d\, R_t}$$

---

## Шаг 5 — Оптимальный порог из Section 22 (Peskir & Shiryaev)

Параметры задачи: $\mu = \delta$, $\sigma = 1$, $\gamma = \delta^2/2$, $\lambda$ (интенсивность разладки), $c$ (стоимость задержки).

**Вспомогательная функция:**

$$\alpha(\pi) = \log\!\frac{\pi}{1-\pi} - \frac{1}{\pi}$$

**Функция для smooth-fit уравнения:**

$$\psi(\pi) = -\frac{c}{\gamma}\,e^{-\frac{\lambda}{\gamma}\alpha(\pi)}\int_0^\pi \frac{e^{\frac{\lambda}{\gamma}\alpha(\rho)}}{\rho\,(1-\rho)^2}\,d\rho$$

**Оптимальный порог** $A_*$ — единственный корень:

$$\psi(A_*) = -1$$

*Существование и единственность:* $\psi(0^+) = 0$, $\psi(\pi) \to -\infty$ при $\pi \to 1$, $\psi$ монотонно убывает.

---

## Шаг 6 — Оптимальное правило остановки

$$\boxed{\tau^* = \inf\!\left\{t \geq 1 : \pi_t \geq A_*\right\}}$$

Эквивалентно: $\tau^* = \inf\!\left\{t : R_t \geq h^*\right\}$, где $h^* = A_* \,/\, [\lambda_d\,(1 - A_*)]$.

---

## Итоговая схема

```
ОФФЛАЙН (калибровка):
──────────────────────────────────────────────────────
Корпус {X_t} ~ p_0  →  обучить RNN_φ + ε_θ
Калибровочная выборка  →  μ_0, σ_0
Задать: δ_min, λ, c  →  γ = δ_min² / 2
Найти A* из ψ(A*) = -1  [бисекция на (0,1)]

ОНЛАЙН (детекция):
──────────────────────────────────────────────────────
Получен X_t
  ↓
h_{t-1} = RNN_φ(X_{t-l}, ..., X_{t-1})
  ↓
ε ~ N(0, I)
L_t = ‖ε_θ(μ(τ)X_t + σ(τ)ε, τ, h_{t-1}) − ε‖²
Z_t = (L_t − μ_0) / σ_0          ←  ~N(0,1) под p_0,  ~N(δ,1) под p_1
  ↓
Λ_t = exp(δ_min · Z_t − δ_min²/2)
R_t = (1 + R_{t-1}) · Λ_t
π_t = λ_d R_t / (1 + λ_d R_t)
  ↓
если π_t ≥ A*  →  ТРЕВОГА
```

---

## Ключевой вывод

Ценой сжатия $X_t \in \mathbb{R}^d \;\to\; Z_t \in \mathbb{R}$ мы получаем **точную** постановку задачи Ширяева с явно известными:

$$p_0 = \mathcal{N}(0,\,1), \qquad p_1 = \mathcal{N}(\delta,\,1)$$

Весь байесовский аппарат Section 22 (Peskir & Shiryaev, 2006) применяется без каких-либо приближений. Оптимальный порог $A_*$ минимизирует байесовский риск:

$$\rho(\tau) = c\,\mathbb{E}\!\left[(\tau - \theta)^+\right] + \mathbb{P}(\tau < \theta)$$


# Универсальна ли редукция к $\mathcal{N}(0,1)$?

---

## Что действительно универсально

По **построению** калибровки:

$$\mathbb{E}[Z_t \mid H_0] = 0, \qquad \mathrm{Var}(Z_t \mid H_0) = 1$$

Это тавтология — $\mu_0$ и $\sigma_0$ оцениваются из данных, поэтому любую скалярную статистику можно так нормировать.

По **MSE-декомпозиции**: сдвиг $\delta > 0$ при $p_1 \neq p_0$ универсален. Модель, обученная на $p_0$, всегда делает бо́льшую ошибку на $p_1$:

$$\mathbb{E}_{p_1}\!\left[\|\hat\varepsilon_\theta - \varepsilon\|^2\right] = \underbrace{\mathbb{E}_{p_1}\!\left[\|\hat\varepsilon_1 - \varepsilon\|^2\right]}_{\geq 0} + \underbrace{\mathbb{E}_{p_1}\!\left[\|\hat\varepsilon_\theta - \hat\varepsilon_1\|^2\right]}_{> 0 \text{ при } p_1 \neq p_0}$$

---

## Что НЕ универсально

**Гауссовость $Z_t$ под $H_0$** требует CLT, а CLT требует приближённой независимости ошибок по размерностям:

$$L_t = \sum_{i=1}^d \left(\hat\varepsilon_{\theta,i} - \varepsilon_i\right)^2$$

Если $\hat\varepsilon_\theta$ сильно коррелирует ошибки между компонентами (что легко при общем энкодере), CLT не работает и $Z_t$ может быть не нормальной.

**Гауссовость под $H_1$ с той же дисперсией** — самое сильное и уязвимое место:

$$Z_t \mid H_1 \approx \mathcal{N}(\delta,\, 1)$$

Дисперсия $L_t$ под $p_1$ вообще говоря не равна $\sigma_0^2$. Если разладка меняет не среднее, а хвосты или структуру корреляций — $\mathrm{Var}(Z_t \mid H_1) \neq 1$, и формула для $\Lambda_t$ неверна.

---

## Итоговая таблица

| Утверждение | Статус |
|---|---|
| $\mathbb{E}[Z_t \mid H_0] = 0$, $\;\mathrm{Var}(Z_t \mid H_0) = 1$ | Точно (по калибровке) |
| $Z_t \mid H_0 \sim \mathcal{N}(0,1)$ | Приближённо (CLT при больших $d$, слабая корреляция ошибок) |
| $\mathbb{E}[Z_t \mid H_1] = \delta > 0$ | Точно (MSE-декомпозиция) |
| $\mathrm{Var}(Z_t \mid H_1) = 1$ | **Приближённо, может нарушаться** |
| $Z_t \mid H_1 \sim \mathcal{N}(\delta, 1)$ | **Приближённо, зависит от типа аномалии** |

---

## Вывод

Редукция к Section 22 **точна** только при выполнении $\mathcal{N}(0,1)$ vs $\mathcal{N}(\delta,1)$. На практике:

- **Хорошо работает** при разладке типа «сдвиг распределения», большом $d$ с приближённо независимыми компонентами ошибок, и хорошо обученной модели.
- **Ломается** при разладке типа «изменение дисперсии или хвостов», сильных межкомпонентных корреляциях в $\hat\varepsilon_\theta$, или переобученной модели.

Подход правильнее рассматривать как **параметрическую аппроксимацию**: CLT даёт гауссовость, а оптимальность Section 22 — дополнительный бонус при выполнении условий. Для надёжности необходима эмпирическая проверка на hold-out:


# Дизайн модели для точной редукции к $\mathcal{N}(0,1)$

---

## Проблема, которую нужно решить

Два места, где редукция приблизительная:

1. **$Z_t \mid H_0 \sim \mathcal{N}(0,1)$** — нарушается, если ошибки денойзера по размерностям коррелированы (CLT не работает)
2. **$\mathrm{Var}(Z_t \mid H_1) = 1$** — нарушается, потому что дисперсия $L_t$ зависит от $p_1$

---

## Идея 1 — Конформная калибровка (без переобучения, точная по $H_0$)

По **обменимости** (exchangeability): если $L_1^{(0)}, \ldots, L_N^{(0)}, L_t^{(0)}$ i.i.d., то ранг $L_t$ в
калибровочной выборке строго равномерен:

$$p_t = \frac{\#\{i : L_i^{(0)} \leq L_t\} + 1}{N + 1} \sim \mathrm{Uniform}(0,1) \quad \text{точно}$$

Преобразуем:

$$Z_t = \Phi^{-1}(p_t) \sim \mathcal{N}(0,1) \quad \textbf{точно, без CLT}$$

Под $H_1$: $L_t$ стохастически больше $\Rightarrow$ $p_t$ больше $\Rightarrow$ $Z_t > 0$ в среднем.

**Ограничение:** сторона $H_1$ определяется рангом, а не сдвигом. Формула $\mathcal{N}(\delta, 1)$
теряет точность — но $H_0$ гарантирована.

---

## Идея 2 — Логарифмический transform + регуляризация корреляций

Если остатки $r_i = \varepsilon_i - \hat\varepsilon_{\theta,i}$ приближённо i.i.d. с дисперсией $\sigma_\perp^2$, то:

$$L_t = \sum_{i=1}^d r_i^2 \sim \sigma_\perp^2\,\chi^2(d)$$

**Ключевое свойство логарифма:** для $L \sim \sigma^2 \chi^2(d)$

$$\mathrm{Var}(\log L) \approx \frac{2}{d} \quad \textbf{не зависит от } \sigma^2$$

Это означает: $\log L_t$ **дисперсионно стабилен** — дисперсия одинакова под $H_0$ и $H_1$.

**Предлагаемая нормировка** (приближение Уилсона–Хилферти):

$$Z_t = \sqrt{\frac{d}{2}} \cdot \log\frac{L_t}{d\,\hat\sigma_\perp^2}$$

По теореме о $\chi^2$: $Z_t \xrightarrow{d} \mathcal{N}(0,1)$ быстрее, чем линейная нормировка при малых $d$.

**Регуляризатор декорреляции остатков** (добавляется в функцию потерь):

$$\mathcal{L}_{\mathrm{white}} = \left\|\hat\Sigma_r - I\right\|_F^2, \qquad
\hat\Sigma_r = \frac{1}{B}\sum_{b=1}^B r^{(b)}{r^{(b)}}^\top$$

Штрафует за корреляции между размерностями $\Rightarrow$ CLT работает лучше,
$\chi^2(d)$-приближение точнее.

**Итоговая функция потерь:**

$$\mathcal{L} = \mathcal{L}_{\mathrm{DSM}} + \lambda\,\mathcal{L}_{\mathrm{white}}$$

---

## Идея 3 — Архитектура с факторизованными остатками

Заставить последний слой денойзера быть **поканальным** (channel-independent):

$$\hat\varepsilon_\theta(x_\tau, \tau, \mathbf{h}) = \mathrm{Head}_{\mathrm{indep}}\!\left(f_\theta(x_\tau, \tau, \mathbf{h})\right)$$

где $\mathrm{Head}_{\mathrm{indep}}$ обрабатывает каждую размерность независимо
(без cross-attention, без MLP по всем каналам одновременно).

Тогда остатки $r_1, \ldots, r_d$ условно независимы при данном $\mathbf{h}$ $\Rightarrow$
CLT **структурно оправдан**, а не только как аппроксимация.

---

## Сравнение подходов

| | **Конформная** | **Log + whitening** | **Факторизованная** |
|---|---|---|---|
| Переобучение нужно? | Нет | Да ($\mathcal{L}_{\mathrm{white}}$) | Да (архитектура) |
| $H_0 = \mathcal{N}(0,1)$ | **Точно** | Приближённо лучше | Приближённо лучше |
| $\mathrm{Var}(Z_t\|H_1) = 1$? | Нет | **Да** ($\log$ убирает $\sigma$-зависимость) | Частично |
| Потеря выразительности? | Нет | Малая | Да (факторизация) |
| Вычислительная цена | Минимальная | Малая | Средняя |

---

## Рекомендуемый дизайн (три слоя)

**Слой 1 — обучение:** декоррелируем остатки

$$\mathcal{L} = \mathcal{L}_{\mathrm{DSM}} + \lambda\,\|\hat\Sigma_r - I\|_F^2$$

**Слой 2 — статистика:** стабилизируем дисперсию

$$Z_t = \sqrt{\frac{d}{2}}\cdot\log\frac{L_t}{d\,\hat\sigma_\perp^2}$$

**Слой 3 — калибровка:** гарантируем $H_0$ без предположений

$$Z_t \leftarrow \Phi^{-1}\!\left(\frac{\#\{L_i^{(0)} \leq L_t\} + 1}{N+1}\right)$$

Каждый слой исправляет отдельную аппроксимацию:
слой 1 делает CLT обоснованным, слой 2 стабилизирует дисперсию при $H_1$,
слой 3 даёт точную гарантию на $H_0$.


# Калиброванная диффузионная модель (Calibrated Residual Diffusion)

---

## Ключевая идея

В стандартном DSM модель учится предсказывать условное матожидание $\hat\varepsilon_\theta = \mathbb{E}[\varepsilon \mid x_\tau, \mathbf{h}]$,
но **ковариация остатков**

$$\mathrm{Cov}(r) = \mathrm{Var}(\varepsilon \mid x_\tau, \mathbf{h}) \neq I$$

игнорируется. Именно поэтому статистика $L_t = \|r\|^2$ не имеет известного
распределения. Если научить модель предсказывать и ковариацию, то расстояние
Махаланобиса

$$T_t = r_t^\top \hat\Sigma_\theta^{-1} r_t$$

будет следовать $\chi^2(d)$ **по построению** — без CLT и без предположений о
независимости компонент.

---

## Архитектура

Два выхода из одного энкодера $f_\theta(x_\tau, \tau, \mathbf{h})$:

$$\hat\varepsilon_\theta \in \mathbb{R}^d \quad \text{(среднее)}, \qquad
\hat\Sigma_\theta \succ 0 \quad \text{(ковариация остатков)}$$

**Параметризация $\hat\Sigma_\theta$ через разложение Холецкого:**

$$\hat\Sigma_\theta = L_\theta L_\theta^\top + \epsilon I, \qquad L_\theta \in \mathbb{R}^{d \times d} \text{ нижнетреугольная}$$

Слагаемое $\epsilon I$ гарантирует положительную определённость.

**Практически:** для больших $d$ используем **низкоранговое приближение** ($k \ll d$):

$$\hat\Sigma_\theta = \hat\sigma_\perp^2 I + U_\theta U_\theta^\top, \qquad
U_\theta \in \mathbb{R}^{d \times k}$$

Первое слагаемое — изотропный фон, второе — $k$ направлений с повышенной неопределённостью.
Матрично-детерминантная лемма даёт $\det$ и обратную матрицу за $O(dk^2)$.

---

## Функция потерь

Отрицательный логарифм правдоподобия многомерного гауссиана
(**собственная оценочная функция**, proper scoring rule):

$$\boxed{
\mathcal{L}(\theta) = \mathbb{E}_{\tau,\, x_0,\, \varepsilon}\!\left[
r^\top \hat\Sigma_\theta^{-1} r + \log\det\hat\Sigma_\theta
\right], \qquad r = \varepsilon - \hat\varepsilon_\theta}$$

**Почему это правильно:**

- Первое слагаемое — нормированная ошибка (MSE в метрике $\hat\Sigma_\theta^{-1}$)
- Второе слагаемое — штраф за завышение ковариации
- При минимуме: $\hat\Sigma_\theta \to \mathrm{Var}(\varepsilon \mid x_\tau, \mathbf{h})$ (истинная апостериорная ковариация)
- При $\hat\Sigma_\theta = I$ совпадает со стандартным DSM — обобщение без противоречий

---

## Тестовая статистика и её распределение

**Определение:**

$$\boxed{T_t = r_t^\top \hat\Sigma_\theta(x_\tau, \tau, \mathbf{h}_{t-1})^{-1}\, r_t,
\qquad r_t = \varepsilon - \hat\varepsilon_\theta(x_\tau, \tau, \mathbf{h}_{t-1})}$$

**Теорема (квадратичная форма гауссиана).**
Если $r \sim \mathcal{N}(0, \Sigma)$ и $\hat\Sigma = \Sigma$, то:

$$T_t \sim \chi^2(d) \quad \textbf{точно, для любого конечного } d$$

Это **не** CLT — распределение $\chi^2(d)$ следует из свойства гауссовой квадратичной формы.
Никаких предположений о независимости компонент не нужно.

**Под $H_1$** (после разладки, $X_t \sim p_1$):
сдвиг среднего $\Delta\mu$ даёт нецентральное хи-квадрат распределение:

$$T_t \sim \chi^2(d,\, \lambda), \qquad \lambda = \Delta\mu^\top \hat\Sigma_\theta^{-1} \Delta\mu > 0$$

Параметр нецентральности $\lambda$ — это квадрат расстояния Махаланобиса между $p_0$ и $p_1$.

---

## Итоговая схема редукции

Из $T_t \sim \chi^2(d)$ под $H_0$ получаем **два варианта** тестовой статистики с известным распределением:

**Вариант A — нормальное приближение** (для больших $d$):

$$Z_t = \frac{T_t - d}{\sqrt{2d}} \xrightarrow{d \to \infty} \mathcal{N}(0, 1)$$

**Вариант B — точное равномерное** (для любого $d$, через CDF):

$$U_t = F_{\chi^2(d)}(T_t) \sim \mathrm{Uniform}(0, 1) \quad \text{точно}$$

Вариант B предпочтителен при малых $d$: нет приближения, можно применить
стандартное конформное преобразование $\Phi^{-1}(U_t) \sim \mathcal{N}(0,1)$.

---

## Применение к задаче Ширяева

Параметры задачи автоматически определяются через модель:

$$\mu = \sqrt{\lambda} = \sqrt{\Delta\mu^\top \hat\Sigma_\theta^{-1} \Delta\mu}, \qquad \sigma = 1$$

Log-likelihood ratio одного наблюдения (нецентральный vs. центральный $\chi^2$):

$$\ell_t = \log \frac{f_{\chi^2(d,\lambda)}(T_t)}{f_{\chi^2(d)}(T_t)}$$

или в гауссовом приближении через $Z_t = (T_t - d)/\sqrt{2d}$:

$$\ell_t \approx \sqrt{\frac{\lambda}{2d}}\cdot Z_t - \frac{\lambda}{4d}$$

Далее применяется статистика Ширяева–Робертса и порог $A_*$ из Section 22
(Peskir & Shiryaev) с $\mu_\mathrm{eff} = \sqrt{\lambda / 2d}$.

---

## Что даёт эта конструкция

| Свойство | Стандартный DSM | Calibrated Residual Diffusion |
|---|---|---|
| Функция потерь | MSE | Гауссовский NLL (proper scoring rule) |
| Ковариация остатков | Игнорируется | Предсказывается явно |
| Распределение $T_t$ под $H_0$ | Неизвестно | $\chi^2(d)$ **по построению** |
| Нужен CLT? | Да | **Нет** |
| Нужна декорреляция? | Да (противоречие) | **Нет** |
| Распределение под $H_1$ | Неизвестно | $\chi^2(d, \lambda)$ (нецентральное) |
| Точный LR доступен? | Нет | **Да** |


## Проблема DSM-эквивалентности для ковариации

---

### Почему DSM работает для среднего

Ключевой шаг — убрать неизвестное $\varepsilon_{post} = \mathbb{E}[\varepsilon \mid x_\tau, \mathbf{h}]$
через закон полного ожидания:

$$\mathbb{E}_{p_\tau}\!\left[\|\hat\varepsilon_\theta - \varepsilon_{post}\|^2\right]
= \mathbb{E}_{x_0,\varepsilon}\!\left[\|\hat\varepsilon_\theta - \varepsilon\|^2\right]
+ \underbrace{\mathbb{E}_{p_\tau}\!\left[\|\varepsilon - \varepsilon_{post}\|^2\right]}_{\text{const w.r.t. }\theta}$$

Это работает потому что $\mathbb{E}[\varepsilon \mid x_\tau] = \varepsilon_{post}$ и перекрёстные члены
сокращаются. Одной пары $(x_0, \varepsilon)$ достаточно.

---

### Почему для ковариации так не получается

Обозначим $r = \varepsilon - \hat\varepsilon_\theta$,
$\hat\Sigma_{post} = \mathrm{Var}(\varepsilon \mid x_\tau, \mathbf{h})$. Разложим:

$$\hat\Sigma_{post}
= \underbrace{\mathbb{E}[rr^\top \mid x_\tau]}_{\text{оценимо}}
- \underbrace{(\hat\varepsilon_\theta - \varepsilon_{post})(\hat\varepsilon_\theta - \varepsilon_{post})^\top}_{\text{bias — неизвестен}}$$

Подставляя в NLL-потерю:

$$\mathbb{E}_{p_\tau}\!\left[\mathrm{tr}(\hat\Sigma_\theta^{-1}\hat\Sigma_{post})\right]
= \mathbb{E}_{x_0,\varepsilon}\!\left[r^\top \hat\Sigma_\theta^{-1} r\right]
- \underbrace{\mathbb{E}_{p_\tau}\!\left[
  \|\hat\Sigma_\theta^{-1/2}(\hat\varepsilon_\theta - \varepsilon_{post})\|^2
\right]}_{\neq\,\text{const},\;\text{зависит от }\hat\varepsilon_\theta}$$

Этот **bias-член содержит $\varepsilon_{post}$ и не убирается** аналогом DSM-трюка.

---

### Решение: двухфазное обучение

**Фаза 1 — стандартный DSM** (обучаем $\hat\varepsilon_\theta$, $\hat\Sigma_\phi$ заморожена):

$$\mathcal{L}_1(\theta)
= \mathbb{E}_{x_0,\varepsilon}\!\left[\|\hat\varepsilon_\theta(x_\tau, \tau, \mathbf{h}) - \varepsilon\|^2\right]$$

После сходимости: $\hat\varepsilon_\theta \approx \varepsilon_{post}$, поэтому **bias $\approx 0$**.

**Фаза 2 — оценка ковариации** (замораживаем $\theta$, обучаем $\hat\Sigma_\phi$):

$$\mathcal{L}_2(\phi)
= \mathbb{E}_{x_0,\varepsilon}\!\left[
  r^\top \hat\Sigma_\phi^{-1} r + \log\det\hat\Sigma_\phi
\right], \qquad r = \varepsilon - \hat\varepsilon_\theta$$

Поскольку $\hat\varepsilon_\theta \approx \varepsilon_{post}$:

$$\mathbb{E}[rr^\top \mid x_\tau] \approx \hat\Sigma_{post}$$

Минимум $\mathcal{L}_2$ достигается при $\hat\Sigma_\phi = \mathbb{E}[rr^\top \mid x_\tau]$ —
**точная оценка апостериорной ковариации** без противоречия с DSM.

---

### Альтернатива: ковариация через якобиан

Апостериорная ковариация связана с гессианом score через обобщённую формулу Твиди:

$$\mathrm{Var}(\varepsilon \mid x_\tau, \mathbf{h})
\approx I + \frac{\sigma(\tau)}{\mu(\tau)}\,\nabla_{x_\tau}\hat\varepsilon_\theta(x_\tau, \tau, \mathbf{h})$$

То есть $\hat\Sigma_\phi$ можно **не обучать отдельно** — вычислять аналитически
через якобиан уже обученного $\hat\varepsilon_\theta$ за один backward pass.

---

### Итог

| | Среднее $\hat\varepsilon_\theta$ | Ковариация $\hat\Sigma_\phi$ |
|---|---|---|
| DSM-эквивалентность | Точная | Приближённая (при малом bias) |
| Что нужно | $\varepsilon$ (один сэмпл) | $\nabla_{x_\tau}\hat\varepsilon_\theta$ или несколько сэмплов |
| Решение | Стандартный DSM | Фаза 2 или якобиан |


## Сравнение: коварицационная голова vs. аппроксимация Твиди

---

### Аппроксимация Твиди (якобиан)

$$\hat\Sigma \approx I + \frac{\sigma(\tau)}{\mu(\tau)}\,\nabla_{x_\tau}\hat\varepsilon_\theta$$

**Плюсы:**
- Не требует дообучения вообще
- Теоретически обоснована

**Минусы:**
- Якобиан $\nabla_{x_\tau}\hat\varepsilon_\theta \in \mathbb{R}^{d \times d}$ — это $d$ backward pass-ов, $O(d^2)$ памяти
- При $d = 500$ (снапшот стакана) — **практически неприменимо** в реальном времени
- Можно считать только диагональ через Хатчинсона, но тогда теряется корреляционная структура
- Аппроксимация, а не точное равенство

---

### Ковариационная голова (фаза 2)

$$\hat\Sigma_\phi = \hat\sigma^2 I + U_\phi U_\phi^\top, \quad U_\phi \in \mathbb{R}^{d \times k},\; k \ll d$$

**Плюсы:**
- На инференсе — **один forward pass** через лёгкую голову, никаких backward
- Low-rank параметризация: $O(dk)$ параметров и вычислений вместо $O(d^2)$
- Детерминант и обратная матрица дёшевы через лемму Вудбери:

$$(\hat\sigma^2 I + UU^\top)^{-1} = \frac{1}{\hat\sigma^2}\left(I - U({\hat\sigma^2 I + U^\top U})^{-1}U^\top\right)$$

- Обучается на правильном критерии (NLL), а не косвенной аппроксимации

**Минусы:**
- Требует фазы 2 обучения
- Качество зависит от того, насколько хорошо сошлась фаза 1

---

### Вывод

| | **Твиди/якобиан** | **Ков. голова** |
|---|---|---|
| Дообучение | Не нужно | Нужна фаза 2 |
| Инференс | $O(d)$ backward pass | $O(dk)$ forward pass |
| Применимость при больших $d$ | Нет | **Да** |
| Корреляционная структура | Полная (но дорого) | Low-rank (дёшево) |
| Точность | Аппроксимация | Обучается напрямую |

**Рекомендация: ковариационная голова** с low-rank $k \sim 10{-}20$.

Якобиан оправдан только если модель нельзя трогать совсем и $d$ мал.
